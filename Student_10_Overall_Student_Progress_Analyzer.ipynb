## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_10_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
print(" Dataset shape:",df.shape)
print("column names :",df.columns.tolist())
df.info()
print("\nMissing values in Each column:\n",df.isnull().sum())
print("\n Total Duplicate rows:", df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# 1. Select the numeric columns and convert them to a NumPy array
marks = df[["Marks_Percent", "Attendance_Percent", "Assignment_Percent", "Quiz_Percent"]].to_numpy()

# 2. Calculate mean, max, and min across the array (axis=0 calculates per column)
print("Mean per column:", np.mean(marks, axis=0))
print("Max per column:", np.max(marks, axis=0))
print("Min per column:", np.min(marks, axis=0))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# Calculated Column 1: Overall average across all four metrics
df["Overall_Average"] = (
    df["Marks_Percent"] 
    + df["Attendance_Percent"] 
    + df["Assignment_Percent"] 
    + df["Quiz_Percent"]
) / 4

# Calculated Column 2: Pass/Fail status based on a 50% threshold
def check_result(score):
    if score >= 50:
        return "Pass"
    else:
        return "Fail"

df["Result"] = df["Overall_Average"].apply(check_result)

# Preview the updated dataframe
df.head()
```

## 6. Your individual project challenge
**Challenge:** Calculate weighted score (40%, 20%, 20%, 20%); identify students below 50 overall.

Write your solution below and explain what the result means.


```python
# Step 1: Calculate the weighted score (40% Marks, 20% Attendance, 20% Assignment, 20% Quiz)
df["Weighted_Score"] = (df["Marks_Percent"] * 0.40) + (df["Attendance_Percent"] * 0.20) + (df["Assignment_Percent"] * 0.20) + (df["Quiz_Percent"] * 0.20)

# Step 2: Filter for students whose weighted score is below 50%
below_50 = df[df["Weighted_Score"] < 50]

# Step 3: Print the total count of students below 50%
print("Total students below 50%:", len(below_50))

# Step 4: Display the filtered table of students below 50%
below_50
```
